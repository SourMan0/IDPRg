```python
import pandas as pd41
```

```python
losses1 = pd.read_csv('../losses/unirepLosses1.csv')
losses2 = pd.read_csv('../losses/unirepLosses2.csv')
losses3 = pd.read_csv('../losses/unirepLosses3.csv')
losses4 = pd.read_csv('../losses/unirepLosses4.csv')
losses5 = pd.read_csv('../losses/unirepLosses5.csv')
# compute mean RMSE across the loaded CSVs and save a combined CSV
dfs = [losses1, losses2, losses3, losses4, losses5]
conc = pd.concat(dfs, keys=['1', '2', '3', '4', '5'])
mean_rmse = conc.groupby(level=1)['RMSE Score'].mean()
# use the first row per original index as template for the other columns (they are the same across CSVs)
template = conc.groupby(level=1).first()
template['RMSE Score'] = mean_rmse
combined_mean = template.reset_index(drop=True)
combined_mean.to_csv('../losses/unirep_mean_losses.csv', index=False)
print(f"Saved 'losses/unirep_mean_losses.csv' with {len(combined_mean)} rows.")
```

```python
# Combine rows (by original row index) choosing the row with the smallest RMSE Score across the loaded CSV dataframes
dfs = [losses1, losses2, losses3, losses4, losses5]

# concat with a key per source so we can identify rows from each dataframe
conc = pd.concat(dfs, keys=['1', '2', '3', '4', '5'])

# for each original row-index (level=1), find the tuple (source, orig_index) with minimal RMSE Score
best_idx = conc.groupby(level=1)['RMSE Score'].idxmin()

# select those best rows, sort by the original index, and reset to a normal RangeIndex
best = conc.loc[best_idx].reset_index(level=0, drop=True).sort_index()
final = best.reset_index(drop=True)

# save final combined CSV
final.to_csv('../losses/unirep_best_losses.csv', index=False)

print(f"Saved '../losses/unirep_best_losses.csv' with {len(final)} rows.")

# compute mean RMSE across the loaded CSVs and save a combined CSV
mean_rmse = conc.groupby(level=1)['RMSE Score'].mean()
# use the first row per original index as template for the other columns (they are the same across CSVs)
template = conc.groupby(level=1).first()
template['RMSE Score'] = mean_rmse
combined_mean = template.reset_index(drop=True)
combined_mean.to_csv('../losses/unirep_mean_losses.csv', index=False)
print(f"Saved '../losses/unirep_mean_losses.csv' with {len(combined_mean)} rows.")
```
